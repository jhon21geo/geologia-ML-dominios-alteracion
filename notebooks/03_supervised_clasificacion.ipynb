La variable objetivo es `MOD_ALT`. Las features son geoquímica Z-score. Compara el perfil `thesis` (Tablas 19–22) con `robust`. Equivale a la **Fase 2** de [GeoIA](https://geoia.site/dominios-ml/) (lienzo Test & Score / matriz / ROC).

```python
from pathlib import Path
import pandas as pd
from alteration_ml.preprocess import prepare_feature_matrix
from alteration_ml.supervised import split_labeled, train_models
from alteration_ml.evaluate import evaluate_model, metrics_table, per_domain_table

df = pd.read_csv(Path("../data/synthetic/synthetic_merged.csv"))
ready, z_cols, _ = prepare_feature_matrix(df, feature_kind="chemistry")
split = split_labeled(ready, z_cols, test_size=0.2, random_state=42)
models = train_models(split, profile="thesis")
results = {name: evaluate_model(m, split.X_test, split.y_test) for name, m in models.items()}
metrics_table(results)
```
