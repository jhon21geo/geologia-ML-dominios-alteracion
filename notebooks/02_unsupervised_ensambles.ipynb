# 02 — PCA, dendrograma y K-Means

Aprendizaje no supervisado sobre scores SWIR. `k=5` se toma del corte del dendrograma de minerales, como en la tesis.

```python
from pathlib import Path
import pandas as pd
from alteration_ml.preprocess import prepare_feature_matrix
from alteration_ml.unsupervised import fit_pca, fit_kmeans, mineral_hierarchical_clusters, cluster_mineral_profiles
from alteration_ml.constants import SPECTRAL_MINERALS

df = pd.read_csv(Path("../data/synthetic/synthetic_merged.csv"))
ready, z_cols, _ = prepare_feature_matrix(df, feature_kind="spectral")
X = ready[z_cols].to_numpy()
pca, scores, var = fit_pca(X, n_components=5)
print("Varianza por PC:", var.round(3))
print("Acumulada PC1+PC2:", var[:2].sum().round(3))
```

```python
z, mineral_table = mineral_hierarchical_clusters(ready, n_clusters=5)
display(mineral_table)
_, labels, sil = fit_kmeans(X, n_clusters=5)
ready["cluster_k5"] = labels
print("Silhouette:", round(sil, 3))
cluster_mineral_profiles(ready, "cluster_k5", list(SPECTRAL_MINERALS))
```
