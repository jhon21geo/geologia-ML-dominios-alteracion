# 01 — EDA espectral y geoquímica

Exploración del yacimiento **sintético** que replica los seis dominios y los 13 minerales SWIR de Mallma (2026). No hay datos de la unidad original.

```python
from pathlib import Path
import pandas as pd
from alteration_ml.synthetic import write_synthetic_tables
from alteration_ml.spectral import spectral_summary, assemblage_flags
from alteration_ml.constants import DOMAINS, DOMAIN_LABELS

data_dir = Path("../data/synthetic")
merged = data_dir / "synthetic_merged.csv"
if not merged.exists():
    write_synthetic_tables(data_dir)
df = pd.read_csv(merged)
df.head()
```

```python
print(df.groupby("MOD_ALT").size().reindex(DOMAINS))
print("\nEtiquetadas:", int(df.labeled.sum()), "/", len(df))
spectral_summary(df)
```

```python
flags = assemblage_flags(df)
pd.crosstab(df.MOD_ALT, flags.ensamble_argavd)
```
