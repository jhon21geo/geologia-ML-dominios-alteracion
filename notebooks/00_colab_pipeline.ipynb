```python
import os, sys, subprocess, pathlib

REPO = "https://github.com/jhon21geo/geologia-ML-dominios-alteracion.git"
BRANCHES = ["cursor/metodologia-tesis-sintetica-fd6d", "main"]

root = pathlib.Path.cwd()
if (root / "src" / "alteration_ml").exists():
    print("Ya estás dentro del repositorio.")
else:
    dest = root / "geologia-ML-dominios-alteracion"
    if not dest.exists():
        cloned = False
        for branch in BRANCHES:
            try:
                subprocess.check_call(["git", "clone", "--depth", "1", "-b", branch, REPO, str(dest)])
                print("Clonado rama", branch)
                cloned = True
                break
            except subprocess.CalledProcessError:
                continue
        if not cloned:
            raise RuntimeError("No se pudo clonar el repositorio.")
    os.chdir(dest)
    print("Directorio:", pathlib.Path.cwd())

subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "-e", "."])
print("Paquete instalado.")
```

```python
from IPython.display import display, Markdown
from alteration_ml.pipeline import run_pipeline

display(Markdown("## Ejecutando el flujo (dato sintético, perfil tesis)"))
summary = run_pipeline(data_dir="data/synthetic", output_dir="outputs", profile="thesis")
display(Markdown(
    f"**Muestras:** {summary['n_samples']} · **etiquetadas:** {summary['n_labeled']} · "
    f"**mejor modelo:** `{summary['best_model']}`"
))
display(summary["ranking"])
```
